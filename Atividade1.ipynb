```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

```

```python

nome_arquivo = 'quantizacao.png'

img = cv2.imread(nome_arquivo)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(5,5))
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem em tons de cinza')
plt.axis('off')
plt.show()
```

```python
def quantizar(img_gray, bits):
    niveis = 2 ** bits
    passo = 256 / niveis
    img_quant = (np.floor(img_gray / passo) * passo).astype(np.uint8)
    return img_quant

img_1bit = quantizar(img_gray, 1)   # 2 níveis
img_2bit = quantizar(img_gray, 2)   # 4 níveis
img_4bit = quantizar(img_gray, 4)   # 16 níveis
img_8bit = quantizar(img_gray, 8)   # 256 níveis
```

```python
fig, ax = plt.subplots(1, 5, figsize=(20, 5))

ax[0].imshow(img_gray, cmap='gray', vmin=0, vmax=255)
ax[0].set_title('Original (tons de cinza)')
ax[0].axis('off')

ax[1].imshow(img_1bit, cmap='gray', vmin=0, vmax=255)
ax[1].set_title('1 bit (2 níveis)')
ax[1].axis('off')

ax[2].imshow(img_2bit, cmap='gray', vmin=0, vmax=255)
ax[2].set_title('2 bits (4 níveis)')
ax[2].axis('off')

ax[3].imshow(img_4bit, cmap='gray', vmin=0, vmax=255)
ax[3].set_title('4 bits (16 níveis)')
ax[3].axis('off')

ax[4].imshow(img_8bit, cmap='gray', vmin=0, vmax=255)
ax[4].set_title('8 bits (256 níveis)')
ax[4].axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png', dpi=150)
plt.show()
```

À medida que reduzimos o número de bits, diminuem os níveis de intensidade e alguns detalhes da imagem são perdidos. As transições ficam mais bruscas, principalmente com 1 bit, que possui apenas preto e branco. Com 8 bits, há 256 níveis de intensidade, preservando melhor os detalhes e as transições suaves.
