```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files


uploads = files.upload()
imagem = cv2.imread("brazuca.jpg")




imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)


imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)



def quantizar(imagem, bits):
    niveis = 2 ** bits


    passo = 256 / niveis


    imagem_quantizada = np.floor(imagem / passo) * passo


    imagem_quantizada = np.clip(imagem_quantizada, 0, 255)

    return imagem_quantizada.astype(np.uint8)



q1 = quantizar(imagem_cinza, 1)
q2 = quantizar(imagem_cinza, 2)
q4 = quantizar(imagem_cinza, 4)
q8 = quantizar(imagem_cinza, 8)



fig, axs = plt.subplots(1, 5, figsize=(20, 5))


axs[0].imshow(imagem_cinza, cmap="gray", vmin=0, vmax=255)
axs[0].set_title("Original\nTons de cinza")
axs[0].axis("off")


axs[1].imshow(q1, cmap="gray", vmin=0, vmax=255)
axs[1].set_title("1 bit\n2 níveis")
axs[1].axis("off")


axs[2].imshow(q2, cmap="gray", vmin=0, vmax=255)
axs[2].set_title("2 bits\n4 níveis")
axs[2].axis("off")


axs[3].imshow(q4, cmap="gray", vmin=0, vmax=255)
axs[3].set_title("4 bits\n16 níveis")
axs[3].axis("off")


axs[4].imshow(q8, cmap="gray", vmin=0, vmax=255)
axs[4].set_title("8 bits\n256 níveis")
axs[4].axis("off")


plt.tight_layout()


plt.savefig("quantizacao.png", dpi=300, bbox_inches="tight")


plt.show()
```
